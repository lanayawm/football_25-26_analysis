```python
import pandas as pd
import mysql.connector
from sqlalchemy import create_engine
```

```python
engine = create_engine('mysql+mysqlconnector://root:1111@localhost:3306/football')
query = 'select * from football.player_data_25_26'
df = pd.read_sql(query, engine)
columns = df.columns.tolist()
```

```python
def clean(colum):
    new_column = []
    for col in colum:
        gol = col.replace('%', '_per')
        new_column.append(gol)
    return new_column
```

```python
new = clean(columns)
df.columns = new
df.to_sql('player_data_25_26', engine, if_exists='replace', index=False)
```
