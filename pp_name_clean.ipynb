```python
import pandas as pd
import numpy as np
import mysql.connector
from sqlalchemy import create_engine
```

```python
engine = create_engine('mysql+mysqlconnector://root:1111@localhost:3306/football')
query = 'select * from football.player_profiles'
df = pd.read_sql(query, engine)
```

```python
for i in range(len(df)):
    row = df.loc[i, 'player_name']
    spot = row.find('(')
    df.loc[i, 'player_name'] = row[0:spot-1]
```

```python
df.to_sql('player_profiles', engine, if_exists='replace', index=False)
```
